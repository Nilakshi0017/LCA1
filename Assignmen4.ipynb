```python
# Addition of two matrices using array

matrix1 = [
    [1, 2, 3],
    [4, 5, 6]
]

matrix2 = [
    [7, 8, 9],
    [10, 11, 12]
]

result = [
    [0, 0, 0],
    [0, 0, 0]
]

# Add the matrices
for i in range(2):
    for j in range(3):
        result[i][j] = matrix1[i][j] + matrix2[i][j]

# Display result
print("Resultant Matrix:")
for row in result:
    print(row)
```
