```python
# Exercise 1: Dictionary Operations

my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34,
    "age": 18,
    "marks": 85
}

print("Original Dictionary:")
print(my_dict)

# Adding student information
my_dict["city"] = "Pune"

print("\nAfter Adding Information:")
print(my_dict)

# Deleting student information
del my_dict["age"]

print("\nAfter Deleting Information:")
print(my_dict)
```

```python
# Exercise 2: Convert List to Dictionary

list1 = ['name', 'panel', 'rollno']
list2 = ['ABC', 'B', 34]

my_dict = dict(zip(list1, list2))

print("Dictionary:")
print(my_dict)
```

```python
# Exercise 3: Sort Dictionary

my_dict = {
    'name': 'ABC',
    'panel': 'B',
    'rollno': 34,
    'marks': [65, 87, 67, 94]
}

print("Original Dictionary:")
print(my_dict)

print("\nSorted Elements:")
print(sorted(my_dict))
```

```python
# Exercise 4: Dictionary to Lists

my_dict = {
    'name': 'ABC',
    'panel': 'B',
    'rollno': 34,
    'marks': [65, 87, 67, 94]
}

keys_list = list(my_dict.keys())
values_list = list(my_dict.values())

print("List of Keys:")
print(keys_list)

print("\nList of Values:")
print(values_list)
```

```python
# Exercise 5: Mean of Dictionary Values

mydict = {
    'marks1': 23,
    'marks2': 123,
    'marks3': 43,
    'marks4': 13,
    'marks5': 39
}

total = sum(mydict.values())
mean = total / len(mydict)

print("Mean of all values:", mean)
```

```python
# Exercise 6: Dictionary with Lists

my_dict = {
    'name': ['Yash', 'Neel', 'Dev'],
    'rollno': [11, 12, 13],
    'marks': [78, 56, 98]
}

# a) Display name as Dev
print("Name:", my_dict['name'][2])

# b) Display 12 roll no
print("Roll No:", my_dict['rollno'][1])

# c) Display greatest marks
print("Greatest Marks:", max(my_dict['marks']))
```

```python
# Exercise 7: Frequency of Letters

text = input("Enter a string: ")

frequency = {}

for letter in text:
    if letter != " ":
        frequency[letter] = frequency.get(letter, 0) + 1

print("Frequency of letters:")
print(frequency)
```

```python

# Exercise 8: Quotient and Remainder

numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

quotient = numerator // denominator
remainder = numerator % denominator

result = (quotient, remainder)

print("Tuple of Quotient and Remainder:")
print(result)
```

```python

# Exercise 9: Number of Days Between Two Dates

from datetime import date

d1 = int(input("Enter day of first date: "))
m1 = int(input("Enter month of first date: "))
y1 = int(input("Enter year of first date: "))

d2 = int(input("Enter day of second date: "))
m2 = int(input("Enter month of second date: "))
y2 = int(input("Enter year of second date: "))

date1 = (d1, m1, y1)
date2 = (d2, m2, y2)

date_object1 = date(y1, m1, d1)
date_object2 = date(y2, m2, d2)

days = abs((date_object2 - date_object1).days)

print("First Date Tuple:", date1)
print("Second Date Tuple:", date2)
print("Number of days between two dates:", days)
```
